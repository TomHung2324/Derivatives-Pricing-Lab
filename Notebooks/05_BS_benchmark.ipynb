# Question 2(a): Black-Scholes Analytic Benchmark

This section evaluates a PLTR call option using the Black-Scholes model.

The analysis uses the market snapshot collected on October 9, 2026, with a selected expiration date of January 15, 2027.

The objectives are to:

1. Calculate the theoretical European call price and delta.
2. Infer the corresponding European put price using put-call parity.
3. Compare the inferred put value with the observed market put bid-ask spread.
4. Explain the risk-neutral pricing assumptions and the relevance of early exercise.

```python
import numpy as np
import pandas as pd
from scipy.stats import norm

s0 = 209.05
K = 210
r = 0.045

valuation_date = pd.Timestamp("2026-10-09")
expiration_date = pd.Timestamp("2027-01-15")

T = (expiration_date - valuation_date).days / 365

# Replace with the full-precision call IV from Q1
sigma = 0.497988

print("Time to maturity:", T)
```

```python
d1 = (np.log(s0/K)+(r+0.5*sigma**2)*T)/(sigma*np.sqrt(T))
d2 = d1 - sigma*np.sqrt(T)

bs_call = s0*norm.cdf(d1)-K*np.exp(-r*T)*norm.cdf(d2)
bs_delta = norm.cdf(d1)
```

## Put-Call Parity and Market Put Comparison

For a non-dividend-paying stock, European put-call parity is given by:

C-P=S_0-Ke^{-rT}


The theoretical European put value is inferred from the Black-Scholes call price.

This value is compared with the observed put bid-ask spread. Since listed equity puts may be American-style options, differences can also reflect an early-exercise premium.

```python
put_parity = bs_call-s0+K*np.exp(-r*T)

print("BS Call Price:", bs_call)
print("BS Call Delta:", bs_delta)
print("Parity-implied Put Price:", put_parity)
```

## Market Put Bid-Ask Comparison

The theoretical European put price inferred from put-call parity is compared with the observed market put bid-ask spread.

The comparison evaluates whether the parity-implied put price lies within the quoted spread. Any discrepancy must be interpreted with caution because listed American-style puts may carry an early-exercise premium.

```python
put_bid = 20.60    # From the October 9 market snapshot
put_ask = 21.35

put_mid = (put_ask+put_bid)/2
within_spread = put_bid<= put_parity <= put_ask

print("Market Put Midquote:", put_mid)
print("Parity-implied Put:", put_parity)
print("Within Bid-Ask Spread:", within_spread)
print("Difference from Put Midquote:", put_parity - put_mid)
```

### Findings: Black–Scholes Benchmark and Put–Call Parity

Using the selected PLTR option, the Black–Scholes call price is approximately **$22.1750**, with a call delta of **0.5629**.

European put–call parity implies a put value of **$20.6030**, which lies within the observed market put bid–ask spread of **$20.60–$21.35**. The implied put value is approximately **$0.3720 below the market midquote** of $20.9750.

This difference does not establish an arbitrage opportunity. The parity-implied value corresponds to a European put, whereas the listed put is American-style and may include an early-exercise premium. Furthermore, the market midquote is not necessarily an executable transaction price.

Black–Scholes pricing uses the risk-neutral drift rather than the historical physical drift because the option can be replicated and priced through no-arbitrage arguments under the model assumptions.

For a non-dividend-paying stock with a nonnegative interest rate, early exercise of an American call is not optimal. Consequently, the European Black–Scholes call value is applicable to the corresponding American call.
