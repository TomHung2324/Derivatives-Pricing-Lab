```python
import yfinance as yf
import pandas as pd
import numpy as np
from datetime import datetime, timezone

pltr = yf.Ticker("PLTR")

spot_data = pltr.history(period = "1d", interval = "1m")
print(spot_data.tail())
s0 = spot_data["Close"].iloc[-1]
print("Spot price: ", s0)
```

```python
import numpy as np
expiries_to_check = [
    "2026-11-20",
    "2026-12-18",
    "2027-01-15",
    "2027-02-19"
]
available_expiries = set(pltr.options)

for expiry in expiries_to_check:
    if expiry not in available_expiries:
        print(expiry, "Not available")
        continue
    chain = pltr.option_chain(expiry)
    calls = chain.calls
    puts =  chain.puts

    vaild_calls = calls[
        (calls["bid"]>0) &
        (calls["ask"]>0) &
        (calls["ask"] >= calls["bid"])    
    ]

    vaild_puts = puts[
        (puts["bid"]>0) &
        (puts["ask"]>0) &
        (puts["ask"] >= puts["bid"])
    ]

    print(
        expiry,
        "Valid call: ", len(vaild_calls),
        "Valid puts: ", len(vaild_puts)
    )

strike_lower_than_spot = calls[
        (calls["bid"]>0) &
        (calls["ask"]>0) &
        (calls["ask"] >= calls["bid"]) &
        (calls["strike"] < s0)    
    ]
print(len(strike_lower_than_spot))

strike_higher_than_spot = calls[
        (calls["bid"]>0) &
        (calls["ask"]>0) &
        (calls["ask"] >= calls["bid"]) &
        (calls["strike"] > s0)    
    ]
print(len(strike_higher_than_spot))

distance_to_spot = calls["strike"]-s0
strike_closest_to_spot = np.abs(distance_to_spot).idxmin()
atm_strike = calls.loc[strike_closest_to_spot, "strike"]
print(atm_strike)
   
   
```

```python
import pandas as pd
import numpy as np

calls = pd.read_csv("PLTR_calls_2027-01-15.csv")
puts = pd.read_csv("PLTR_puts_2027-01-15.csv")

print("Calls:", calls.shape)
print("Puts:", puts.shape)

display(calls.head())
display(puts.head())
```

```python
for name, df in [("Calls", calls), ("Puts", puts)]:
    valid = (
        (df["bid"] > 0)
        & (df["ask"] > 0)
        & (df["ask"] >= df["bid"])
    )

    print(name)
    print("Total contracts:", len(df))
    print("Valid quotes:", valid.sum())
    print("Valid quote ratio:", valid.mean())
```

```python
s0 = 198.72

for df in [calls, puts]:

    df["mid"] = (df["bid"] + df["ask"]) / 2

    df["spread"] = df["ask"] - df["bid"]

    df["relative_spread"] = (
        df["spread"] / df["mid"].replace(0, np.nan)
    )

    df["moneyness"] = s0 / df["strike"]

atm_calls = calls[
    (calls["strike"] >= 180)
    & (calls["strike"] <= 220)
]

atm_puts = puts[
    (puts["strike"] >= 180)
    & (puts["strike"] <= 220)
]

columns = [
    "strike", "bid", "ask", "mid",
    "relative_spread", "volume",
    "openInterest", "impliedVolatility"
]

display(atm_calls[columns])
display(atm_puts[columns])
```

```python
atm_call = calls.loc[calls["strike"] == 200].iloc[0]
atm_put = puts.loc[puts["strike"] == 200].iloc[0]

print("Call last trade:", atm_call["lastTradeDate"])
print("Put last trade:", atm_put["lastTradeDate"])

print("Call IV:", atm_call["impliedVolatility"])
print("Put IV:", atm_put["impliedVolatility"])

print("Call Mid:", atm_call["mid"])
print("Put Mid:", atm_put["mid"])
```

```python
from pathlib import Path

data_dir = Path("../Data")
data_dir.mkdir(parents=True, exist_ok=True)

for expiry in expiries_to_check:
    chain = pltr.option_chain(expiry)

    expiry_calls = chain.calls.copy()
    expiry_puts = chain.puts.copy()

    expiry_calls["expiration"] = expiry
    expiry_puts["expiration"] = expiry

    expiry_calls.to_csv(
        data_dir / f"PLTR_calls_{expiry}.csv",
        index=False
    )

    expiry_puts.to_csv(
        data_dir / f"PLTR_puts_{expiry}.csv",
        index=False
    )

    print(expiry, len(expiry_calls), len(expiry_puts))
```
